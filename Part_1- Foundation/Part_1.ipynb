## 1. What is Machine Learning?

**Machine Learning (ML)** is all about teaching computers to learn from data, just like humans learn from experience.

Instead of programming every rule manually, we:

- Give machines lots of data
- Let them find patterns
- Use those patterns to make predictions or decisions

### Traditional Programming vs Machine Learning

| Traditional Programming | Machine Learning |
|---|---|
| Give rules + data → get result | Give data + result → get rules/model |
| Human writes the logic | Machine learns patterns automatically |

### Real-World Examples

- YouTube recommendations → learns from watch history
- Spam detection in Gmail → learns patterns in spam emails
- Voice assistants → learn how you speak
- Self-driving cars → learn to identify signs, pedestrians and roads
- Face unlock → learns to recognize your face
- Netflix → predicts what you may want to watch

# 2. AI vs ML vs DL

AI, Machine Learning and Deep Learning are related but different.

## Artificial Intelligence — AI

AI is the **big umbrella**.

> AI is the science of making machines smart, similar to human intelligence.

### Examples

- Playing chess like a human
- Talking to Alexa
- Driving a car on its own

**AI = Any system that mimics human intelligence.**

---

## Machine Learning — ML

ML is a **subset of AI**.

Machines learn from data and improve over time.

### Examples

- YouTube recommending videos
- Netflix predicting your next binge
- Gmail filtering spam

**ML = Machines learning patterns from data.**

---

## Deep Learning — DL

DL is a **subset of Machine Learning**.

It uses **neural networks**, which are inspired by the human brain.

### Examples

- Face recognition on phones
- ChatGPT
- Self-driving car vision

**DL = ML using neural networks for big and complex data such as images or speech.**

### Relationship

```text
Artificial Intelligence
        │
        └── Machine Learning
                │
                └── Deep Learning
```

# 3. Types of Machine Learning

Machine Learning has 3 major types:

1. Supervised Learning
2. Unsupervised Learning
3. Reinforcement Learning

---

# 4. Supervised Learning

Supervised Learning is like a **student being taught by a teacher**.

We give the machine:

- Input
- Correct output

The machine learns the relationship between them and then predicts the output for new data.

### Basic Idea

```text
Input Data + Correct Output
          ↓
      ML Algorithm
          ↓
      Trained Model
          ↓
 New Input → Prediction
```

### Example

Suppose we want to predict whether someone will get a loan.

| Income | Credit Score | Loan |
|---:|---:|---|
| 40,000 | 750 | Yes |
| 25,000 | 600 | No |
| 50,000 | 800 | Yes |
| 30,000 | 580 | No |

Here:

- Income = Input variable
- Credit Score = Input variable
- Loan = Output variable

The model learns from known examples and predicts the loan decision for a new person.

### Algorithms

- Linear Regression
- Logistic Regression
- Decision Tree
- Naive Bayes
- Support Vector Machine (SVM)
- K-Nearest Neighbors (KNN)
- Gradient Boosting
- AdaBoost

---

# 5. Unsupervised Learning

In **Unsupervised Learning**, we have data but **no labelled/correct output**.

The machine tries to discover:

- Patterns
- Groups
- Relationships
- Structure
- Complexities/anomalies

### Basic Idea

```text
Unlabelled Data
      ↓
ML Algorithm
      ↓
Patterns / Groups / Structure
```

It can be used for:

- Finding patterns
- Grouping similar items
- Reducing complexity
- Spotting outliers

Unsupervised learning can also be used with supervised learning to make models.

---

# 6. Reinforcement Learning

Reinforcement Learning is like **training a dog**.

The system learns through **trial and error**.

- Good behaviour → Reward
- Bad behaviour → Punishment

### Basic Idea

```text
Agent
  ↓
Takes Action
  ↓
Environment
  ↓
Reward / Punishment
  ↓
Learns Better Action
```

### Key Idea

> **Learning by trial and error.**

# 7. EDA — Exploratory Data Analysis

### Full Form

**EDA = Exploratory Data Analysis**

EDA is the process of **visually and statistically exploring and understanding your data before building a Machine Learning model.**

It helps us:

- Understand the data
- Discover patterns
- Spot anomalies
- Generate insights
- Decide what to do next

### Simple Definition

> EDA is like being a detective — you're looking at your data with curiosity before building models or doing transformations.

# 9. EDA Steps

## Step 1 — Viewing the Data

First understand what your dataset looks like.

Useful things to check:

- `head()`
- `tail()`
- `shape`
- `columns`
- Data types

Questions to ask:

- What columns do I have?
- What types of data are present?

---

## Step 2 — Summary Statistics

Check:

- Mean
- Median
- Mode
- Standard deviation
- Minimum
- Maximum
- Quartiles

These help us understand:

- Central tendency
- Spread
- Distribution of data

---

## Step 3 — Value Counts

Ask:

> How many unique values are present in a column?

This is especially useful for **categorical columns**.

---

## Step 4 — Missing Value Analysis

Ask:

> Where are the gaps? What percentage of the data is missing?

Missing values need to be identified before training the model.

---

## Step 5 — Visualizations

### Histograms

Show the **distribution of values**.

### Boxplots

Help identify:

- Outliers
- Spread

### Bar Plots

Used for comparisons of categories.

### Correlation Heatmaps

Show **linear relationships between numerical features**.

### Scatter Plots

Show **bivariate relationships** between two variables.

---

## Step 6 — Target Variable Exploration

Ask:

> How does your output, like "charges" in your dataset, relate to other variables?

The **target variable** is the variable that we ultimately want to predict.

---

# 10. Importance of EDA

EDA is important because:

- You can't clean or process what you don't understand.
- It helps identify mistakes, biases and limitations in the data.
- It guides the direction of data cleaning and feature engineering.
- It gives the audience/stakeholders an overview of what the data is telling you.

---

# 11. Data Cleaning

Data cleaning means identifying and fixing problems in the dataset.

There are 7 major areas.

---

## 1. Handle Missing Values

First check which columns have missing values (nulls).

### Strategies

- Drop missing rows/columns — only if very few are missing
- Impute missing values

### For Numerical Data

- Mean
- Median

### For Categorical Data

- Mode

### Advanced Methods

- Linear Regression
- KNN
- Interpolation

---

## 2. Remove Duplicates

Detect and remove **exact duplicate rows**.

```text
Duplicate Data
      ↓
   Identify
      ↓
    Remove
```

---

## 3. Fix Data Types

Convert incorrect data types.

Example:

Numbers stored as strings should be converted into numerical data.

---

## 4. Handle Inconsistent Categories

Clean categorical values so that the same category has one standard format.

Example:

```text
"Male"
"male"
"MALE"
```

should become:

```text
"Male"
```

Similarly:

```text
"Yes"
"yes"
"YES"
```

should use one standard format.

---

## 5. Detect and Handle Outliers

Outliers are unusually high or low observations.

### Methods

- Boxplots
- IQR
- Z-score

### Ways to Handle Them

- Remove them if clearly wrong
- Cap them, e.g. at the 95th percentile

---

## 6. Fix Logic or Domain Errors

Check whether values make sense according to the domain.

Examples:

```text
Age = -5
BMI = 150
```

These may indicate errors.

Possible solutions:

- Replace with mean
- Replace with median
- Remove the value

---

## 7. Overall Data Cleaning Idea

> **EDA tells you what's wrong. Data Cleaning fixes it.**

> **Cleaning is not glamorous, but it's 80% of the work in real-world projects.**

# 12. Data Preprocessing

# 13. Encoding Categorical Variables

Machine Learning models generally need numerical representations.

Therefore, categorical variables need to be converted into numbers.

There are two common methods:

1. Label Encoding
2. One-Hot Encoding

---

## 1. Label Encoding

Used for **ordered categories**.

Example:

```text
Low    → 1
Medium → 2
High   → 3
```

There is a meaningful order.

---

## 2. One-Hot Encoding

Used for **non-ordered categories**.

Example:

```text
Region:
North
South
East
West
```

Each category gets its own binary column.

# 15. Feature Scaling

Feature scaling brings numerical values to the **same scale**.

It is especially useful for **distance-based algorithms**.

Two important methods are:

1. Normalization
2. Standardization

---

## Normalization — Min-Max Scaling

Scales values between:

**0 and 1**

### Formula

\[
X_{new} = \frac{X-X_{min}}{X_{max}-X_{min}}
\]

---

## Standardization — Z-Score Scaling

Transforms data so that it has:

- Mean = 0
- Standard deviation = 1

### Formula

\[
Z = \frac{X-\mu}{\sigma}
\]

Where:

- `X` = Original value
- `μ` = Mean
- `σ` = Standard deviation

# 16. Feature Engineering

### Definition

Feature Engineering means:

> **Creating new features or transforming existing features to expose useful patterns that ML models can use.**

### Why Do We Need It?

Because:

> **ML models don't know domain logic — we have to give them the right signals.**

### Common Feature Engineering Techniques

- Mathematical combinations
- Target-based flags
- Time-based features, if time exists

### Example

From:

```text
Date of Birth
```

we can create:

```text
Age
```

---

# 17. Feature Selection

Feature Selection means:

> **Selecting the most useful features and removing the rest.**

### Importance

It:

- Reduces noise
- Reduces overfitting
- Speeds up training
- Improves model accuracy
- Makes model interpretation easier

---

# 18. Feature Selection Methods

There are three broad methods:

1. Filter Methods
2. Wrapper Methods
3. Embedded Methods

---

## 1. Filter Methods — Pure Statistics

These methods use statistical relationships.

### Correlation Matrix

Used to identify and remove **highly correlated features**.

### Chi-Square Test

Used for:

**Categorical vs Categorical**

### ANOVA F-Test

Used for:

**Numerical vs Categorical Target**

---

## 2. Wrapper Methods

Wrapper methods select features by evaluating different feature subsets with a machine learning model.

The detailed wrapper-method explanation is not included in the available notes.

---

## 3. Embedded Methods

Feature selection happens **during model training**.

### Lasso Regression

Shrinks coefficients toward zero.

### Tree-Based Models

Examples:

- Random Forest
- XGBoost

These provide **feature importance scores**.
