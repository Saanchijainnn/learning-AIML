# CLASSIFICATION ALGORITHMS — MACHINE LEARNING NOTES

## 1. Introduction to Classification

Classification is a **supervised machine learning technique** used to predict the category or class of a given input.

The model learns from labelled training data and predicts the class of new data.

**Examples:**
- Email classification: Spam or Not Spam
- Disease prediction: Disease or No Disease
- Student result: Pass or Fail
- Image classification: Cat, Dog, or Bird

**Types of Classification:**
1. Binary Classification: Two classes, e.g., Yes/No.
2. Multiclass Classification: More than two classes, e.g., Apple, Mango, Orange.
3. Multilabel Classification: One input can belong to multiple classes simultaneously.

---

## 2. Logistic Regression

### Definition

Logistic Regression is a supervised machine learning algorithm primarily used for classification. It predicts the probability that an input belongs to a particular class.

Although its name contains the word *regression*, it is commonly used for classification problems.

### Linear Regression vs Logistic Regression

| Linear Regression | Logistic Regression |
|---|---|
| Predicts continuous values. | Predicts class probabilities. |
| Output can be any real number. | Output lies between 0 and 1. |
| Uses a straight-line equation. | Uses the sigmoid function to convert the linear output into a probability. |
| Example: Predicting house prices. | Example: Predicting spam emails. |

### Step 1: Linear Equation

The algorithm first calculates a linear combination of the input features.

$$
z = w_1x_1 + w_2x_2 + \cdots + w_nx_n + b
$$

Where:

- \(x_1, x_2, \ldots, x_n\) = input features
- \(w_1, w_2, \ldots, w_n\) = weights
- \(b\) = bias
- \(z\) = linear output

### Step 2: Sigmoid Function

The sigmoid function converts the linear output into a probability between 0 and 1.

$$
\sigma(z) = \frac{1}{1 + e^{-z}}
$$

The predicted probability is:

$$
p = \sigma(z) = \frac{1}{1 + e^{-z}}
$$

Where \(e\) is Euler's number, approximately 2.718.

**Example:**

If the model predicts \(p = 0.85\), it estimates an 85% probability of belonging to Class 1.

### Step 3: Classification Threshold

A threshold is used to convert the predicted probability into a class label. A common default threshold is 0.5.

$$
\hat{y} =
\begin{cases}
1, & \text{if } p \geq 0.5 \\
0, & \text{if } p < 0.5
\end{cases}
$$

Example:
- Probability = 0.85 → Class 1
- Probability = 0.25 → Class 0

The threshold can be changed depending on the problem.

### Step 4: Log Loss

Log Loss measures the difference between actual labels and predicted probabilities. A lower Log Loss indicates better predictions.

For one observation:

$$
L = -\left[y\ln(p) + (1-y)\ln(1-p)\right]
$$

For \(m\) observations:

$$
J = -\frac{1}{m}\sum_{i=1}^{m}
\left[
y_i\ln(p_i) + (1-y_i)\ln(1-p_i)
\right]
$$

Where:
- \(y\) = actual class (0 or 1)
- \(p\) = predicted probability of Class 1
- \(m\) = number of observations

**Understanding Log Loss:**
- Actual class = 1, predicted probability = 0.99 → Low loss.
- Actual class = 1, predicted probability = 0.01 → High loss.

### Advantages
1. Simple and easy to implement.
2. Produces probability estimates.
3. Computationally efficient.
4. Works well for many binary classification problems.

### Limitations
1. May not perform well when the decision boundary is highly nonlinear.
2. Can be affected by multicollinearity.
3. Feature engineering may be needed for complex relationships.

## 3. Confusion Matrix

### Definition

A Confusion Matrix is a table used to evaluate a classification model by comparing actual classes with predicted classes.

For binary classification:

|  | Predicted Positive | Predicted Negative |
|---|---|---|
| Actual Positive | True Positive (TP) | False Negative (FN) |
| Actual Negative | False Positive (FP) | True Negative (TN) |

### Four Important Terms

**1. True Positive (TP):** The model predicts positive, and the actual class is positive.

Example: A sick patient is correctly identified as sick.

**2. True Negative (TN):** The model predicts negative, and the actual class is negative.

Example: A healthy patient is correctly identified as healthy.

**3. False Positive (FP):** The model predicts positive, but the actual class is negative.

Example: A healthy patient is incorrectly classified as sick.

**4. False Negative (FN):** The model predicts negative, but the actual class is positive.

Example: A sick patient is incorrectly classified as healthy.

### Evaluation Metrics

#### 1. Accuracy

Accuracy measures the proportion of all predictions that are correct.

$$
\text{Accuracy} = \frac{TP + TN}{TP + TN + FP + FN}
$$

#### 2. Precision

Precision measures how many of the observations predicted as positive are actually positive.

$$
\text{Precision} = \frac{TP}{TP + FP}
$$

**Example:** In spam detection, precision is important when we want to avoid incorrectly marking genuine emails as spam.

#### 3. Recall (Sensitivity)

Recall measures how many of the actual positive observations the model correctly identifies.

$$
\text{Recall} = \frac{TP}{TP + FN}
$$

**Example:** In disease detection, recall is important because missing a sick patient can be dangerous.

#### 4. F1 Score

The F1 Score is the harmonic mean of precision and recall.

$$
F1 = 2 \times
\frac{\text{Precision} \times \text{Recall}}
{\text{Precision} + \text{Recall}}
$$

It is useful when both precision and recall matter, especially with imbalanced datasets.

**Note:** If a metric's denominator is zero, its value needs to be handled appropriately by the evaluation software.

## 4. K-Nearest Neighbors (KNN)

### Definition

K-Nearest Neighbors (KNN) is a supervised machine learning algorithm that predicts the class of a new observation by examining the classes of its nearest training observations.

It is a distance-based algorithm and is often called a lazy learning algorithm because it does not build a conventional model during training.

### Working of KNN

1. Choose the value of \(K\).
2. Calculate the distance between the new observation and the training observations.
3. Sort the observations by distance.
4. Select the \(K\) nearest neighbors.
5. For classification, count their class labels.
6. Assign the majority class to the new observation.

### Euclidean Distance

Euclidean distance measures the straight-line distance between two points.

For two points in \(n\)-dimensional space:

$$
d(X,Y) = \sqrt{\sum_{i=1}^{n}(x_i-y_i)^2}
$$

For two-dimensional points:

$$
d(X,Y) = \sqrt{(x_1-y_1)^2 + (x_2-y_2)^2}
$$

Where:
- \(X\) and \(Y\) = two observations
- \(x_i\) and \(y_i\) = their feature values
- \(d(X,Y)\) = distance between them

### Example

Suppose \(K = 3\), and the three nearest neighbors have these labels:

- Neighbor 1: Cat
- Neighbor 2: Dog
- Neighbor 3: Cat

The majority class is Cat, so KNN predicts **Cat**.

### Choosing K

- A very small \(K\) can make the model sensitive to noise.
- A very large \(K\) can make the model overlook local patterns.
- An odd value of \(K\) can help avoid ties in binary classification, but does not guarantee that ties are impossible.
- Cross-validation can help select a suitable value of \(K\).

### Feature Scaling

Feature scaling is important for KNN because features with larger numerical ranges can dominate the distance calculation.

### Advantages
1. Simple to understand and implement.
2. No complex training process.
3. Can model nonlinear decision boundaries.
4. Can be used for classification and regression.

### Limitations
1. Prediction can be slow for large datasets.
2. Sensitive to irrelevant features and feature scales.
3. Performance can decrease in high-dimensional data.
4. Requires an appropriate choice of \(K\).

## 5. Naive Bayes Classifier

### Definition

Naive Bayes is a supervised classification algorithm based on Bayes' Theorem. It assumes that the features are conditionally independent of one another given the class.

This independence assumption makes the algorithm simple and computationally efficient.

### Bayes' Theorem

$$
P(A \mid B) = \frac{P(B \mid A)\,P(A)}{P(B)}
$$

Where:

- \(P(A \mid B)\) = posterior probability
- \(P(B \mid A)\) = likelihood
- \(P(A)\) = prior probability
- \(P(B)\) = evidence

### Naive Bayes Formula

For a class \(C\) and features \(X_1, X_2, \ldots, X_n\):

$$
P(C \mid X_1,\ldots,X_n)
\propto
P(C)\prod_{i=1}^{n}P(X_i \mid C)
$$

The predicted class is the class with the highest posterior probability.

$$
\hat{C} = \arg\max_C P(C \mid X)
$$

Here, \(\arg\max\) means selecting the class that gives the highest probability.

### Types of Naive Bayes

1. **Gaussian Naive Bayes:** Used when numerical features are modelled using a Gaussian (normal) distribution.
2. **Multinomial Naive Bayes:** Commonly used for count-based features, such as word counts in text classification.
3. **Bernoulli Naive Bayes:** Used for binary features, such as whether a word appears in a document.

### Example: Spam Detection

Suppose an email contains words such as "prize" and "winner".

Naive Bayes estimates the probability that the email is spam using:
- The prior probability of spam.
- The likelihood of the observed words given spam.
- The corresponding probabilities for other classes.

It then predicts the class with the highest posterior probability.

### Advantages
1. Fast and computationally efficient.
2. Works well for many text-classification tasks.
3. Can work with relatively small training datasets.
4. Supports different types of feature distributions through different variants.

### Limitations
1. The feature-independence assumption may be unrealistic.
2. Zero-frequency probabilities can occur for unseen feature values; smoothing can help.
3. Probability estimates may not always be well calibrated.

## 6. Decision Tree

### Definition

A Decision Tree is a supervised machine learning algorithm that makes predictions by repeatedly splitting data according to feature-based conditions.

It can be used for both classification and regression.

### Structure of a Decision Tree

1. **Root Node:** The topmost node, representing the first split.
2. **Decision Node:** A node where a condition or test is applied.
3. **Branch:** The outcome of a condition.
4. **Leaf Node:** The final prediction or output class.

### Working of a Decision Tree

1. Start with the complete training dataset.
2. Select a feature that creates a useful split.
3. Divide the dataset into subsets based on that feature.
4. Repeat the splitting process for each subset.
5. Stop when a stopping condition is met, such as reaching a maximum depth or having sufficiently pure nodes.
6. Use the leaf node to predict the class of a new observation.

### Entropy

Entropy measures the impurity or uncertainty of a dataset.

$$
H(S) = -\sum_{i=1}^{c}p_i\log_2(p_i)
$$

Where:
- \(S\) = dataset
- \(c\) = number of classes
- \(p_i\) = proportion of observations belonging to class \(i\)

**Interpretation:**
- Entropy = 0 means the node is pure.
- Higher entropy means greater class uncertainty.
- For a binary classification problem, entropy is at most 1 when both classes have equal proportions.

For a binary dataset:

$$
H(S) = -p_1\log_2(p_1)-p_0\log_2(p_0)
$$

### Information Gain

Information Gain measures the reduction in entropy after splitting the dataset using a feature.

$$
IG(S,A) = H(S) -
\sum_{v \in Values(A)}
\frac{|S_v|}{|S|}H(S_v)
$$

Where:
- \(H(S)\) = entropy before splitting
- \(A\) = feature used for splitting
- \(S_v\) = subset produced by outcome \(v\)
- \(|S_v|/|S|\) = proportion of data in that subset

A feature with higher Information Gain produces a greater reduction in entropy.

### Example

Suppose a decision tree predicts whether a person will play outside based on:

- Outlook
- Temperature
- Humidity
- Wind

The algorithm evaluates possible splits and chooses one according to its splitting criterion. It then repeats this process for the resulting branches.

The exact root feature depends on the dataset and the calculated split scores.

### Advantages
1. Easy to understand and interpret.
2. Can handle nonlinear relationships.
3. Requires relatively little feature preprocessing.
4. Can model interactions between features.

### Limitations
1. Deep trees can overfit the training data.
2. Small changes in data can produce a different tree.
3. Greedy splitting may not produce a globally optimal tree.
4. Pruning or depth restrictions may be required to control overfitting.

## 7. Support Vector Machine (SVM)

### Definition

Support Vector Machine (SVM) is a supervised machine learning algorithm that finds a decision boundary, called a **hyperplane**, to separate classes.

SVM aims to find a separating hyperplane with the maximum margin between the classes.

### Important Terms

**1. Hyperplane:** The decision boundary that separates classes.

For a linear SVM, the hyperplane can be written as:

$$
w^T x + b = 0
$$

Where:
- \(w\) = weight vector
- \(x\) = input feature vector
- \(b\) = bias

**2. Margin:** The distance between the decision boundary and the nearest observations from the classes. SVM aims to maximize this margin.

**3. Support Vectors:** The training observations closest to the decision boundary. These observations are especially important in determining the boundary.

### Margin Formula

For a standard linear SVM with decision boundaries:

$$
w^T x + b = 1
$$

and

$$
w^T x + b = -1
$$

the distance between the two margin boundaries is:

$$
\text{Margin} = \frac{2}{\|w\|}
$$

Maximizing the margin is equivalent to minimizing the norm of the weight vector under the standard hard-margin formulation.

### Kernel Trick

A kernel allows SVM to model nonlinear decision boundaries without explicitly computing a high-dimensional feature representation.

Common kernels include:

1. **Linear Kernel:** For approximately linearly separable data.
2. **Polynomial Kernel:** Can represent polynomial relationships.
3. **RBF (Gaussian) Kernel:** Can model more complex nonlinear boundaries.
4. **Sigmoid Kernel:** Another kernel option available in some SVM implementations.

### Advantages
1. Effective in high-dimensional feature spaces.
2. Can model nonlinear boundaries with suitable kernels.
3. Uses support vectors to define the decision boundary.
4. Can perform well when classes have a clear margin of separation.

### Limitations
1. Training can be expensive on large datasets.
2. Kernel and regularization parameters require tuning.
3. Feature scaling is often important.
4. The model may be less interpretable than a small decision tree.

## 8. Quick Revision Table

| Algorithm | Main Idea | Common Use |
|---|---|---|
| Logistic Regression | Predicts class probability using the sigmoid function | Binary classification |
| KNN | Predicts using the nearest observations | Classification and regression |
| Naive Bayes | Uses Bayes' Theorem with a conditional-independence assumption | Spam and text classification |
| Decision Tree | Splits data using feature-based conditions | Classification and regression |
| SVM | Finds a maximum-margin decision boundary | Classification |
