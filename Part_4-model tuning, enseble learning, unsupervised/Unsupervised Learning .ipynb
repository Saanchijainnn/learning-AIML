## 1. Model Tuning (Hyperparameter Tuning)

### Definition
Model Tuning is the process of selecting the best hyperparameter values to improve the performance of a machine learning model.

The first version of a model may not perform well. By adjusting its hyperparameters, we try to improve its ability to generalize to unseen data.

### What are Hyperparameters?

Hyperparameters are settings chosen before or during model training that control how a model learns.

**Examples:**

- KNN: Number of neighbors (\(K\))
- Decision Tree: Maximum depth (`max_depth`)
- Random Forest: Number of trees (`n_estimators`)
- Gradient Boosting: Learning rate
- Ridge and Lasso Regression: Regularization strength (\(\alpha\))

### Goals of Model Tuning

1. Find a suitable combination of hyperparameters.
2. Improve performance on unseen data.
3. Reduce overfitting and underfitting.
4. Build a model that generalizes well.

### Overfitting, Underfitting and Poor Performance

**Overfitting:** The model learns the training data too closely, including noise, and performs poorly on unseen data.

**Underfitting:** The model is too simple to learn the important patterns in the data.

**Poor Performance:** The model does not achieve satisfactory results. This may be caused by unsuitable hyperparameters, poor data quality, inappropriate features or an unsuitable algorithm.

## 2. Cross-Validation

### Definition

Cross-validation is a model evaluation technique that divides the available data into multiple parts to estimate how well a model performs on unseen data.

It is useful for comparing models and selecting hyperparameters.

### K-Fold Cross-Validation

In K-Fold Cross-Validation, the dataset is divided into \(K\) approximately equal folds.

**Steps:**

1. Divide the dataset into \(K\) folds.
2. Use one fold for validation.
3. Use the remaining \(K-1\) folds for training.
4. Repeat the process until every fold has been used for validation once.
5. Calculate the average score across all folds.

### Example: 5-Fold Cross-Validation

Suppose the dataset contains 1000 observations.

Each fold contains approximately 200 observations.

- Fold 1: Validate on Fold 1; train on Folds 2–5.
- Fold 2: Validate on Fold 2; train on Folds 1, 3, 4 and 5.
- Continue until all five folds have been used for validation.

If the validation accuracies are 82%, 81%, 86%, 83% and 88%, the average accuracy is:

$$
\text{Average Accuracy}
= \frac{82+81+86+83+88}{5}
$$

$$
\text{Average Accuracy} = 84\%
$$

### General Formula

For \(K\) folds:

$$
\text{Average Score}
= \frac{1}{K}\sum_{i=1}^{K} S_i
$$

Where \(S_i\) is the evaluation score for fold \(i\).

### Advantages

1. Uses different portions of the data for validation.
2. Gives a more reliable performance estimate than a single split in many situations.
3. Helps compare models and hyperparameters.
4. Helps identify overfitting.

**Important:** Keep the final test set separate from model tuning. For classification, stratified K-fold can help preserve class proportions in each fold.

## 3. Hyperparameter Tuning Methods

Three approaches covered in the lecture are manual search, Grid Search and Randomized Search.

### A. Manual Search

In manual search, the user selects hyperparameter values based on experience and evaluates the model.

**Example:** Try \(K=3\), then \(K=5\), and then \(K=7\) for KNN.

**Limitation:** It can be time-consuming and may miss better combinations.

### B. Grid Search with Cross-Validation

Grid Search evaluates every specified combination of hyperparameter values, usually with cross-validation.

**Example: KNN**

Suppose the parameters are:

- `n_neighbors`: 3, 5, 7
- `weights`: uniform, distance
- `metric`: Manhattan, Euclidean

The total number of combinations is:

$$
N = 3 \times 2 \times 2
$$

$$
N = 12
$$

Thus, Grid Search evaluates 12 combinations. If using 5-fold cross-validation, this normally requires 60 model fits, excluding any additional final refit.

**Advantages:**
1. Systematically evaluates all specified combinations.
2. Easy to understand and reproduce.
3. Can identify the best combination within the supplied grid.

**Limitation:** Large grids can require substantial computation.

### C. Randomized Search

Randomized Search evaluates a fixed number of randomly selected hyperparameter combinations from specified distributions or candidate values.

**Advantages:**
1. Often faster than Grid Search for large search spaces.
2. Can explore a wide range of values.
3. Useful when some hyperparameters have many possible values.

**Limitation:** It may not test the best combination because only a subset of combinations is evaluated.

### Grid Search vs Randomized Search

| Grid Search | Randomized Search |
|---|---|
| Evaluates every specified combination. | Evaluates a selected number of sampled combinations. |
| Can be expensive for large grids. | Usually offers better control over computation. |
| Best when the search space is small. | Useful when the search space is large. |

## 4. Ensemble Learning

### Definition

Ensemble Learning is a machine learning technique that combines predictions from multiple models to produce a final prediction.

The individual models are often called **base models** or **base learners**.

The main idea is that combining models can improve predictive performance compared with using a single model.

### Example

Suppose three models predict whether an observation belongs to Class 1:

- Logistic Regression predicts Class 1.
- Decision Tree predicts Class 0.
- KNN predicts Class 1.

With majority voting, the ensemble predicts Class 1.

### Main Types of Ensemble Learning

1. Bagging
2. Boosting
3. Stacking

---

## 5. Bagging (Bootstrap Aggregating)

### Definition

Bagging trains multiple models on different bootstrap samples of the training dataset and combines their predictions.

Bootstrap sampling means drawing observations from the original training dataset **with replacement**.

### Working of Bagging

1. Start with the training dataset.
2. Create multiple bootstrap samples.
3. Train a separate model on each sample.
4. Collect predictions from all trained models.
5. Combine the predictions to produce the final output.

### Combining Predictions

**For classification:** Use majority voting or an aggregation of predicted probabilities.

For majority voting:

$$
\hat{y} = \operatorname{mode}
\left(\hat{y}_1,\hat{y}_2,\ldots,\hat{y}_M\right)
$$

**For regression:** Take the average of the predictions.

$$
\hat{y} = \frac{1}{M}\sum_{i=1}^{M}\hat{y}_i
$$

Where:

- \(M\) = number of models
- \(\hat{y}_i\) = prediction from model \(i\)
- \(\hat{y}\) = final ensemble prediction

### Advantages

1. Often reduces variance.
2. Can reduce overfitting for unstable models such as decision trees.
3. Models can often be trained independently and in parallel.

### Limitations

1. Requires training multiple models.
2. May increase memory usage and prediction time.
3. Does not always improve performance.

### Random Forest

Random Forest is an ensemble learning algorithm based on bagging.

It trains multiple decision trees using bootstrap samples and introduces additional randomness by considering a subset of features when selecting splits.

The predictions from the trees are combined to obtain the final result.

**Key point:** Random Forest is commonly used for both classification and regression.

---

## 6. Boosting

### Definition

Boosting is an ensemble technique in which models are trained sequentially so that later models focus on errors or patterns that earlier models did not handle well.

Boosting often combines many weak learners to create a stronger predictive model.

### Working of Boosting

1. Train an initial weak learner.
2. Identify or emphasize the observations that the model predicts poorly.
3. Train the next learner to improve the overall prediction.
4. Continue adding learners.
5. Combine the learners to produce the final prediction.

### Types of Boosting

#### A. AdaBoost (Adaptive Boosting)

AdaBoost adjusts the importance or weights of training observations based on previous prediction errors. Subsequent learners focus more on observations that were difficult to classify.

The final prediction combines the weak learners using their assigned weights.

#### B. Gradient Boosting

Gradient Boosting builds models sequentially to improve the current predictions by optimizing a loss function.

A simplified update is:

$$
F_m(x) = F_{m-1}(x) + \eta h_m(x)
$$

Where:

- \(F_m(x)\) = updated model prediction
- \(F_{m-1}(x)\) = previous model prediction
- \(h_m(x)\) = new learner's contribution
- \(\eta\) = learning rate

The exact learner and update depend on the chosen loss function and implementation.

#### C. XGBoost

XGBoost stands for Extreme Gradient Boosting.

It is an optimized gradient boosting implementation that includes regularization and efficient training techniques.

Important hyperparameters include:

- `n_estimators`: Number of boosting rounds or trees.
- `learning_rate`: Controls the contribution of each new learner.
- `max_depth`: Limits the depth of trees.
- `subsample`: Controls the fraction of training observations sampled for a boosting round.

### Advantages of Boosting

1. Can achieve strong predictive performance.
2. Learns from errors made by previous learners.
3. Can model complex relationships.

### Limitations of Boosting

1. Sequential training can be computationally expensive.
2. Sensitive tuning may be required.
3. Can overfit if not appropriately regularized.

## 7. Stacking (Stacked Generalization)

### Definition

Stacking combines predictions from different base models using another model called a **meta-model**.

For example, Logistic Regression, Decision Tree and KNN can act as base models, while another Logistic Regression model or KNN model acts as the meta-model.

### Working of Stacking

1. Train multiple different base models.
2. Generate predictions from the base models.
3. Use those predictions as features for the meta-model.
4. Train the meta-model to learn how to combine the base-model predictions.
5. Use the meta-model to generate the final prediction.

### Example

Suppose the base models generate these predictions:

| Observation | Logistic Regression | Decision Tree | KNN |
|---|---:|---:|---:|
| A | 1 | 0 | 1 |
| B | 0 | 0 | 1 |
| C | 1 | 1 | 1 |

These predictions can be used as inputs to the meta-model.

The meta-model learns how to combine them to make a final prediction.

**Important:** In proper stacking, training predictions for the meta-model should generally be generated using out-of-fold predictions. This reduces information leakage and helps prevent overly optimistic performance estimates.

### Bagging vs Boosting vs Stacking

| Bagging | Boosting | Stacking |
|---|---|---|
| Models are trained on bootstrap samples. | Models are trained sequentially. | Base models feed predictions into a meta-model. |
| Often reduces variance. | Focuses on improving previous predictions. | Learns how to combine different models. |
| Example: Random Forest. | Examples: AdaBoost, Gradient Boosting, XGBoost. | Example: Decision Tree + KNN + Logistic Regression with a meta-model. |

## 8. Unsupervised Learning

### Definition

Unsupervised Learning is a type of machine learning in which the data does not contain labelled target outputs. The model attempts to discover patterns, structures or relationships in the data.

### Goals

1. Group similar observations.
2. Reduce the number of features while retaining important information.
3. Detect anomalies or outliers.
4. Discover associations and similarities.

### Supervised vs Unsupervised Learning

| Supervised Learning | Unsupervised Learning |
|---|---|
| Uses labelled data. | Uses data without target labels. |
| Learns to predict a target. | Learns patterns or structure. |
| Includes classification and regression. | Includes clustering and dimensionality reduction. |
| Example: Disease classification. | Example: Customer segmentation. |

### Common Techniques

**1. Clustering:** Groups similar observations.

Examples: K-Means and DBSCAN.

**2. Dimensionality Reduction:** Reduces the number of features while retaining useful information.

Example: Principal Component Analysis (PCA).

**3. Anomaly Detection:** Identifies unusual observations that differ from typical data patterns.

### Applications

1. Customer segmentation based on spending behaviour.
2. Market and product segmentation.
3. Fraud detection and network security.
4. Finding similar users or products in recommendation systems.
5. Image compression.
6. Discovering groups or patterns in medical data.
7. Reducing features before training supervised models.

## 9. K-Means Clustering

### Definition

K-Means is an unsupervised clustering algorithm that divides data into \(K\) clusters. Each observation is assigned to the cluster with the nearest centroid.

A centroid represents the mean position of the observations assigned to a cluster.

### Working of K-Means

**Step 1: Choose K**

Select the number of clusters to create.

**Step 2: Initialize Centroids**

Choose initial centroid positions, commonly using random initialization or K-Means++.

**Step 3: Assign Observations**

Assign each observation to its nearest centroid using a distance measure such as Euclidean distance.

$$
d(X,C_j) = \sqrt{\sum_{i=1}^{n}(x_i-c_{ji})^2}
$$

Where:

- \(X\) = observation
- \(C_j\) = centroid of cluster \(j\)
- \(n\) = number of features

**Step 4: Update Centroids**

Calculate the mean of the observations assigned to each cluster.

For cluster \(j\):

$$
C_j = \frac{1}{N_j}\sum_{i=1}^{N_j}X_i
$$

Here, \(N_j\) is the number of observations assigned to cluster \(j\).

**Step 5: Repeat**

Repeat the assignment and centroid-update steps until the centroids or assignments stabilize, or a stopping condition is reached.

### Example

Suppose a dataset contains customers' ages and salaries.

K-Means can group customers into clusters with similar age and salary patterns. These clusters may help businesses understand different customer segments.

### Advantages

1. Simple and relatively efficient.
2. Works well for compact, well-separated clusters.
3. Can scale to reasonably large datasets.

### Limitations

1. The value of \(K\) must be chosen in advance.
2. Sensitive to initialization and outliers.
3. Can struggle with irregularly shaped clusters.
4. Feature scaling is often important.
5. Different initializations may produce different results.

## 10. Elbow Method and WCSS

### Definition

The Elbow Method is a technique used to help select a suitable number of clusters \(K\) for K-Means.

It examines how the within-cluster sum of squares changes as the number of clusters increases.

### What is WCSS?

WCSS stands for **Within-Cluster Sum of Squares**. It measures the sum of squared distances between observations and their assigned cluster centroids.

$$
WCSS = \sum_{j=1}^{K}\sum_{X_i \in C_j}
\|X_i-\mu_j\|^2
$$

Where:

- \(K\) = number of clusters
- \(C_j\) = observations in cluster \(j\)
- \(X_i\) = observation
- \(\mu_j\) = centroid of cluster \(j\)

### Working of the Elbow Method

1. Run K-Means for different values of \(K\).
2. Calculate WCSS for each value.
3. Plot \(K\) on the horizontal axis and WCSS on the vertical axis.
4. Look for the point where the rate of decrease in WCSS starts slowing down noticeably.
5. Select that point as a possible value of \(K\).

### Interpretation

- Increasing \(K\) generally decreases WCSS.
- A very large \(K\) can create many small clusters.
- The elbow is a practical heuristic, not a guarantee of the mathematically optimal number of clusters.

## 11. DBSCAN Clustering

### Definition

DBSCAN stands for **Density-Based Spatial Clustering of Applications with Noise**.

It is a density-based clustering algorithm that groups closely packed observations and identifies isolated observations as noise or outliers.

Unlike K-Means, DBSCAN does not require the number of clusters to be specified in advance.

### Important Parameters

**1. Epsilon (\(\varepsilon\))**

The maximum distance used to define the neighbourhood around an observation.

**2. MinPts**

The minimum number of observations required within an epsilon neighbourhood for the point to qualify as a core point.

### Types of Points

**Core Point:** Has at least `MinPts` observations in its epsilon neighbourhood, including itself under the standard convention.

**Border Point:** Has fewer than `MinPts` observations in its own neighbourhood but lies within the neighbourhood of a core point.

**Noise Point:** Cannot be assigned to a cluster under the DBSCAN density-reachability rules.

### Working of DBSCAN

1. Select \(\varepsilon\) and `MinPts`.
2. Examine each observation's neighbourhood.
3. Identify core points.
4. Expand clusters through density-connected points.
5. Mark observations that do not belong to any cluster as noise.

### Advantages

1. Does not require \(K\) to be specified.
2. Can identify clusters of irregular shapes.
3. Can identify noise and outliers.

### Limitations

1. Choosing \(\varepsilon\) and `MinPts` can be difficult.
2. Can struggle when clusters have very different densities.
3. Distance-based behaviour can become less useful in high-dimensional data.
4. Feature scaling may be important.

### K-Means vs DBSCAN

| K-Means | DBSCAN |
|---|---|
| Requires the number of clusters \(K\). | Does not require the number of clusters in advance. |
| Assigns observations to centroid-based clusters. | Groups observations by density. |
| Can struggle with irregular cluster shapes. | Can identify irregularly shaped clusters. |
| Sensitive to outliers. | Can label isolated observations as noise. |
| Requires centroid initialization. | Uses \(\varepsilon\) and `MinPts`. |

## 12. Curse of Dimensionality

### Definition

The Curse of Dimensionality refers to the problems that arise when the number of features or dimensions becomes large.

As dimensionality increases, data can become sparse, distance measures may become less informative, and more observations may be required to discover reliable patterns.

### Problems Caused by High Dimensionality

1. **Difficult Pattern Detection:** Patterns can become harder to identify.
2. **Sparse Data:** Observations occupy a very small portion of the possible feature space.
3. **Distance Problems:** Distance-based algorithms such as KNN and K-Means may become less effective.
4. **Higher Computational Cost:** More features can increase training time.
5. **Higher Storage Requirements:** More feature values require more memory.
6. **Overfitting Risk:** A large number of features can make it easier to fit noise, especially with limited training data.

### Algorithms Affected

Distance-based algorithms, such as KNN and K-Means, can be affected by high dimensionality. DBSCAN can also struggle because neighbourhood distances may become less informative.

The extent of the problem depends on the data, feature representation and algorithm.

### How to Reduce Dimensionality

Two important approaches are:

1. Feature Selection
2. Feature Extraction

---

## 13. Feature Selection

### Definition

Feature Selection is the process of selecting a subset of the original features that are most useful for the task.

It removes unnecessary, redundant or less informative features without creating new features.

### Example

Suppose a dataset contains these features:

- Age
- Salary
- Customer ID
- Purchase Frequency

If Customer ID provides no useful predictive information, it may be removed.

### Common Approaches

1. **Filter Methods:** Select features using statistical measures, such as correlation or mutual information.
2. **Wrapper Methods:** Evaluate feature subsets by training and assessing a model.
3. **Embedded Methods:** Select or penalize features during model training, as in Lasso Regression.

### Advantages

1. Reduces the number of input features.
2. Can reduce training time.
3. Can improve interpretability.
4. May reduce overfitting.

---

## 14. Feature Extraction

### Definition

Feature Extraction transforms the original features into a smaller set of new features that capture useful information from the original data.

Unlike feature selection, it creates a new feature representation.

### Example

Suppose a dataset contains:

- Engine Width
- Engine Height
- Engine Size

A feature extraction method may transform several correlated measurements into fewer components that retain much of the variation in the original data.

### Feature Selection vs Feature Extraction

| Feature Selection | Feature Extraction |
|---|---|
| Selects existing features. | Creates transformed features. |
| Original feature meanings are retained. | New features may be harder to interpret. |
| Example: Removing an irrelevant column. | Example: PCA. |

## 15. Principal Component Analysis (PCA)

### Definition

Principal Component Analysis (PCA) is a dimensionality-reduction technique that transforms correlated numerical features into a smaller set of new features called **principal components**.

The first principal component captures the greatest possible variance among all unit-length linear combinations of the features. Each subsequent component captures the greatest remaining variance while being orthogonal to the previous components.

### Important Terms

**1. Principal Components:** New transformed features created from combinations of the original features.

**2. Variance:** Measures how much a feature or component varies across observations.

**3. Eigenvectors:** Directions that define the principal component axes for the covariance matrix.

**4. Eigenvalues:** Values that indicate the variance captured along the corresponding principal component directions.

### Working of PCA

**Step 1: Standardize the Features**

Features with different scales can dominate the analysis, so standardization is often performed.

For a feature \(X\):

$$
z = \frac{X-\mu}{\sigma}
$$

Where:

- \(X\) = original value
- \(\mu\) = feature mean
- \(\sigma\) = feature standard deviation
- \(z\) = standardized value

**Step 2: Calculate the Covariance Matrix**

The covariance matrix describes how pairs of features vary together.

For a centered data matrix \(X\) with \(n\) observations and \(p\) features:

$$
\Sigma = \frac{1}{n-1}X^T X
$$

This expression assumes that the columns of \(X\) have been centered by subtracting their means.

**Step 3: Calculate Eigenvalues and Eigenvectors**

Find the eigenvalues and eigenvectors of the covariance matrix.

$$
\Sigma v = \lambda v
$$

Where:

- \(\Sigma\) = covariance matrix
- \(v\) = eigenvector
- \(\lambda\) = corresponding eigenvalue

**Step 4: Sort the Principal Components**

Arrange the components in descending order of their eigenvalues.

The component with the largest eigenvalue captures the greatest variance.

**Step 5: Select the Components**

Choose the first \(k\) principal components to retain the desired amount of information.

**Step 6: Transform the Data**

Project the standardized data onto the selected component directions.

$$
Z = XW
$$

Where:

- \(X\) = centered or standardized data matrix
- \(W\) = matrix whose columns are the selected component directions
- \(Z\) = transformed data matrix

### Explained Variance Ratio

The explained variance ratio indicates the fraction of total variance captured by a principal component.

For component \(i\):

$$
EVR_i = \frac{\lambda_i}{\sum_{j=1}^{p}\lambda_j}
$$

For the first \(k\) components, the cumulative explained variance ratio is:

$$
EVR_{\text{cumulative}}
= \frac{\sum_{i=1}^{k}\lambda_i}
{\sum_{j=1}^{p}\lambda_j}
$$

For example, if the first two components explain 90% of the variance, the two-component representation retains 90% of the variance measured in the standardized or centered dataset.

### Advantages of PCA

1. Reduces the number of features.
2. Can reduce storage and computation requirements.
3. Can help visualize high-dimensional data in two or three dimensions.
4. Can reduce redundancy among correlated features.

### Limitations of PCA

1. Some information is lost when components are discarded.
2. Components may be difficult to interpret.
3. PCA captures linear directions of variance and may miss nonlinear structure.
4. Scaling and preprocessing choices affect the result.
5. High explained variance does not guarantee better performance for every prediction task.

### Important Note

PCA does not guarantee that all useful predictive information is retained. The number of components should be selected based on the task and evaluated appropriately.
