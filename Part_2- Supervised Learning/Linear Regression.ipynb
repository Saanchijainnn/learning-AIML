## 1. What is Linear Regression?

**Linear Regression is a Supervised Machine Learning algorithm used to predict a continuous numerical value by finding the relationship between independent and dependent variables.**

Examples:
- Predicting salary based on age or experience.
- Predicting house prices based on area.
- Predicting sales based on advertising expenditure.
- Predicting medical expenses based on patient characteristics.

### Types of Variables

- **Independent Variable (X):** The input feature used for prediction.
- **Dependent Variable (Y):** The output we want to predict.

Example: If we predict salary based on experience:
- X = Years of experience
- Y = Salary

### Types of Linear Regression

1. **Simple Linear Regression:** Uses one independent variable.
2. **Multiple Linear Regression:** Uses two or more independent variables.

---

## 2. Equation of Linear Regression

The equation of a straight line is:

**y = mx + c**

In Machine Learning, it is commonly written as:

**ŷ = θ₀ + θ₁x**

Where:

- **ŷ** = Predicted output
- **x** = Input feature
- **θ₀** = Intercept (bias)
- **θ₁** = Slope (coefficient)

### Meaning of Slope and Intercept

**Slope (θ₁):**
- Represents how much the predicted output changes when X increases by one unit.
- A positive slope indicates an increasing relationship.
- A negative slope indicates a decreasing relationship.

**Intercept (θ₀):**
- Represents the predicted value of Y when X = 0.

### Example

Suppose the equation is:

**Salary = 10,000 + 5,000 × Experience**

For 3 years of experience:

Salary = 10,000 + 5,000 × 3

**Predicted Salary = 25,000**

Here, 10,000 is the intercept and 5,000 is the slope.

---

## 3. Best Fit Line

A **Best Fit Line** is the line that best represents the relationship between the input and output data points.

In a scatter plot, the actual observations may not lie exactly on a straight line. Linear Regression finds a line that minimizes the prediction errors.

### How it works

1. Plot the actual data points.
2. Fit a straight line through the data.
3. Calculate the predicted values.
4. Calculate the errors between actual and predicted values.
5. Find the line that minimizes the overall error.

The most common method is **Ordinary Least Squares (OLS)**, which minimizes the sum of squared residuals.

---

## 4. Residuals and Prediction Error

A **residual** is the difference between the actual value and the predicted value.

**Residual = Actual Value − Predicted Value**

Mathematically:

**eᵢ = yᵢ − ŷᵢ**

Where:
- yᵢ = Actual value
- ŷᵢ = Predicted value
- eᵢ = Residual or error

### Example

Actual salary = ₹30,000

Predicted salary = ₹28,000

Residual = 30,000 − 28,000

**Residual = ₹2,000**

A positive residual means the prediction was lower than the actual value. A negative residual means the prediction was higher.

---

## 5. Cost Function

A **Cost Function** measures how far the model's predictions are from the actual values.

The objective of Linear Regression is to find the best values of θ₀ and θ₁ that minimize the cost function.

### Mean Squared Error (MSE)

MSE calculates the average of the squared prediction errors.

**MSE = (1/n) Σ(yᵢ − ŷᵢ)²**

In Linear Regression, a commonly used cost function is:

**J(θ₀, θ₁) = (1/2n) Σ(ŷᵢ − yᵢ)²**

Where:
- J = Cost function
- n = Number of training examples
- yᵢ = Actual value
- ŷᵢ = Predicted value

The factor 1/2 is included for mathematical convenience when differentiating.

### Why do we minimize the cost function?

- To reduce prediction errors.
- To find the best-fit line.
- To improve model performance.

**Lower cost generally means better fit to the training data.**

---

## 6. Gradient Descent

**Gradient Descent is an optimization algorithm used to find parameter values that minimize the cost function.**

In Linear Regression, it adjusts the intercept and slope to reduce prediction error.

### Working Steps

1. Initialize θ₀ and θ₁ with starting values.
2. Calculate predictions using the current parameters.
3. Calculate the cost function.
4. Compute the gradients (derivatives) of the cost function.
5. Update the parameters.
6. Repeat until the cost stops decreasing significantly or a stopping condition is reached.

### Update Formula

**θⱼ := θⱼ − α × ∂J/∂θⱼ**

Where:
- θⱼ = Model parameter
- α = Learning rate
- J = Cost function
- ∂J/∂θⱼ = Gradient of the cost with respect to the parameter

For simple Linear Regression:

**θ₀ := θ₀ − α × (1/n) Σ(ŷᵢ − yᵢ)**

**θ₁ := θ₁ − α × (1/n) Σ[(ŷᵢ − yᵢ)xᵢ]**

Both parameter updates are calculated using the gradients from the current parameter values before updating them.

### Intuition

Imagine standing on a hill and trying to reach its lowest point. You take steps downhill based on the slope. Gradient Descent follows a similar process to find a lower cost.

---

## 7. Learning Rate

The **Learning Rate (α)** determines the size of each parameter update during Gradient Descent.

### Effects of Learning Rate

**If the learning rate is too small:**
- Training becomes very slow.
- Many iterations may be required.

**If the learning rate is too large:**
- The algorithm may overshoot the minimum.
- The cost may fluctuate or increase.
- It may fail to converge.

**If the learning rate is appropriate:**
- The cost generally decreases steadily.
- The algorithm can converge efficiently.

The learning rate is a hyperparameter chosen before or during model tuning.

## 8. Multiple Linear Regression

**Multiple Linear Regression predicts one continuous dependent variable using two or more independent variables.**

### Formula

**ŷ = θ₀ + θ₁x₁ + θ₂x₂ + ... + θₚxₚ**

Where:
- ŷ = Predicted output
- θ₀ = Intercept
- θ₁, θ₂, ..., θₚ = Coefficients
- x₁, x₂, ..., xₚ = Input features
- p = Number of input features

### Example

Predicting salary using experience, education and skills:

Salary = θ₀ + θ₁(Experience) + θ₂(Education) + θ₃(Skills)

The model estimates how each feature contributes to the predicted salary, holding the other included features constant.

### Simple vs Multiple Linear Regression

| Simple Linear Regression | Multiple Linear Regression |
|---|---|
| Uses one independent variable | Uses two or more independent variables |
| ŷ = θ₀ + θ₁x | ŷ = θ₀ + θ₁x₁ + ... + θₚxₚ |
| Easier to visualize in 2D | More complex to visualize |
| Example: Salary from experience | Example: Salary from experience, education and skills |

## 9. Performance Metrics

Performance metrics help evaluate how well a regression model predicts numerical values.

Important metrics include:

1. R² Score
2. Adjusted R² Score
3. Mean Absolute Error (MAE)
4. Mean Squared Error (MSE)
5. Root Mean Squared Error (RMSE)

### 9.1 R² Score — Coefficient of Determination

R² measures how much of the variation in the target variable is explained by the model relative to a baseline that predicts the target mean.

**Formula:**

R² = 1 − (SSᵣₑₛ / SSₜₒₜ)

Where:
- SSᵣₑₛ = Sum of squared residuals
- SSₜₒₜ = Total sum of squares around the target mean

### Interpretation

- R² = 1: Perfect fit to the evaluated observations.
- R² = 0: No improvement over predicting the target mean, under the standard evaluation setup.
- R² < 0: The model performs worse than that baseline on the evaluated data.

Example: R² = 0.86 means the model explains 86% of the observed variation relative to the mean baseline on that dataset.

**Important:** A high R² does not automatically mean the model generalizes well to new data.

### 9.2 Adjusted R²

Adjusted R² modifies R² by accounting for the number of predictors and the sample size.

**Formula:**

Adjusted R² = 1 − [(1 − R²)(n − 1)/(n − p − 1)]

Where:
- R² = Coefficient of determination
- n = Number of observations
- p = Number of independent variables

### Why use Adjusted R²?

- Ordinary R² does not decrease simply because extra predictors are added to an ordinary least-squares model with an intercept on the same dataset.
- Adjusted R² penalizes the addition of predictors that do not improve the model sufficiently.
- It is useful when comparing regression models with different numbers of predictors, provided they are evaluated on comparable data.

### 9.3 MAE — Mean Absolute Error

MAE measures the average absolute difference between actual and predicted values.

**MAE = (1/n) Σ|yᵢ − ŷᵢ|**

Lower MAE indicates smaller average absolute prediction errors.

### 9.4 MSE — Mean Squared Error

**MSE = (1/n) Σ(yᵢ − ŷᵢ)²**

MSE squares each error, so large errors have a greater influence on the score.

Lower MSE is better when comparing models on the same target and dataset.

### 9.5 RMSE — Root Mean Squared Error

**RMSE = √MSE**

RMSE is expressed in the same units as the target variable, making the size of prediction errors easier to interpret.

Lower RMSE generally indicates better predictive accuracy on the evaluated dataset.

## 10. Overfitting and Underfitting

These are two common problems in Machine Learning.

### 10.1 Overfitting

**Overfitting occurs when a model learns the training data too closely, including noise and random fluctuations, and performs poorly on unseen data.**

Characteristics:
- Very good training performance.
- Poorer testing performance.
- High variance.

Example:

Training accuracy = 92%

Testing accuracy = 68%

The large gap suggests possible overfitting.

### Causes

- Model is too complex.
- Too many irrelevant features.
- Insufficient training data.
- Excessive fitting to noise.

### Solutions

- Use regularization.
- Remove irrelevant features.
- Collect more representative data.
- Use cross-validation.
- Reduce model complexity when appropriate.

### 10.2 Underfitting

**Underfitting occurs when a model is too simple to capture the important patterns in the data.**

Characteristics:
- Poor training performance.
- Poor testing performance.
- High bias.

Example:

Training accuracy = 69%

Testing accuracy = 68%

Both results are poor, suggesting possible underfitting.

### Causes

- Model is too simple.
- Important features are missing.
- Model training is insufficient.
- The relationship is more complex than the model can represent.

### Solutions

- Use a more suitable model.
- Add meaningful features.
- Reduce excessive regularization.
- Improve the training process.

### Comparison

| Overfitting | Underfitting |
|---|---|
| Model learns training data too closely | Model fails to learn enough |
| High variance | High bias |
| Good training performance | Poor training performance |
| Poorer testing performance | Poor testing performance |
| Reduce complexity or regularize | Increase suitable model complexity |

**Note:** The training/testing examples above are illustrative. For regression, compare appropriate regression metrics rather than classification accuracy.

## 11. Ridge Regression

**Ridge Regression is a regularized form of Linear Regression that adds an L2 penalty to the cost function.**

It discourages excessively large coefficient values.

### Cost Function

J(θ) = MSE + λ Σθⱼ²

The penalty is normally applied to the feature coefficients, excluding the intercept.

Where:
- MSE = Mean Squared Error
- λ = Regularization strength
- θⱼ = Feature coefficients

### Features

- Reduces large coefficient values.
- Helps control overfitting.
- Can improve generalization.
- Usually shrinks coefficients toward zero but does not force them to exactly zero.

### Effect of λ

- λ = 0: Equivalent to ordinary least-squares Linear Regression.
- Larger λ: Stronger regularization and greater coefficient shrinkage.

---

## 12. Lasso Regression

**Lasso Regression is a regularized form of Linear Regression that adds an L1 penalty to the cost function.**

### Cost Function

J(θ) = MSE + λ Σ|θⱼ|

The penalty is normally applied to the feature coefficients, excluding the intercept.

### Features

- Reduces large coefficient values.
- Can shrink some coefficients exactly to zero.
- Performs automatic feature selection in this sense.
- Helps control overfitting.

### Ridge vs Lasso Regression

| Ridge Regression | Lasso Regression |
|---|---|
| Uses L2 regularization | Uses L1 regularization |
| Penalizes squared coefficients | Penalizes absolute coefficients |
| Usually does not make coefficients exactly zero | Can make some coefficients exactly zero |
| Retains features but shrinks coefficients | Can eliminate features by setting coefficients to zero |

### Why use Regularization?

Regularization adds a penalty for large coefficients, helping prevent the model from becoming unnecessarily complex.

The regularization strength λ is generally selected using validation or cross-validation.

## 13. Complete Linear Regression Workflow

1. Define the problem.
2. Collect the dataset.
3. Explore the data using EDA.
4. Handle missing values, duplicates and outliers appropriately.
5. Encode categorical features if needed.
6. Split data into training and testing sets.
7. Select the relevant features.
8. Train a Linear Regression model.
9. Make predictions.
10. Evaluate the model using MAE, MSE, RMSE and R².
11. Check for overfitting or underfitting.
12. Tune the model or use Ridge/Lasso regularization if needed.
13. Evaluate the final model on held-out test data.

**Important:** Fit preprocessing steps and model parameters using training data only. Do not use the test set to tune the model.

- **MSE:** Average squared prediction error.
- **RMSE:** Square root of MSE.
- **Overfitting:** Learns training data too closely.
- **Underfitting:** Fails to learn important patterns.

- **Lasso:** Uses L1 regularization and can perform feature selection.

**Understand → Fit Line → Calculate Error → Minimize Cost → Evaluate → Regularize if Needed**

Linear Regression aims to find a suitable relationship between input features and a continuous output while keeping prediction errors low and performance reliable on unseen data.
